```python
from pyspark.sql import SparkSession
from pyspark.ml.clustering import KMeans, BisectingKMeans
from pyspark.ml.evaluation import ClusteringEvaluator

def train_and_evaluate_clustering(model, dataset, model_name):
    """Train a clustering model and evaluate it."""
    # Train the model
    trained_model = model.fit(dataset)
    # Make predictions
    predictions = trained_model.transform(dataset)
    predictions.show(truncate=False)
    # Evaluate using Silhouette score
    evaluator = ClusteringEvaluator()
    silhouette = evaluator.evaluate(predictions)
    print(f"{model_name} Silhouette score with squared euclidean distance: {silhouette}")
    # Print cluster centers
    print(f"{model_name} Cluster Centers:")
    for center in trained_model.clusterCenters():
        print(center)
    return trained_model

# Initialize Spark session
spark = SparkSession.builder.appName("ClusteringExample").getOrCreate()

# Load dataset
dataset = spark.read.format("libsvm").load("kmeans_data.txt")

# KMeans Clustering
kmeans = KMeans().setK(2).setSeed(1)
train_and_evaluate_clustering(kmeans, dataset, "KMeans")

# Bisecting KMeans Clustering
bkm = BisectingKMeans().setK(2).setSeed(1)
train_and_evaluate_clustering(bkm, dataset, "BisectingKMeans")

# Stop the Spark session
spark.stop()
```
