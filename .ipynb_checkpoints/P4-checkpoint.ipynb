4. Implement K-means clustering using MapReduce.

```python
import numpy as np
import matplotlib.pyplot as plt

cities = {
    "New York": (40.7128, -74.0060),
    "Los Angeles": (34.0522, -118.2437),
    "Chicago": (41.8781, -87.6298),
    "Houston": (29.7604, -95.3698),
    "Phoenix": (33.4484, -112.0740),
    "San Francisco": (37.7749, -122.4194),
    "Seattle": (47.6062, -122.3321),
    "Boston": (42.3601, -71.0589),
    "Miami": (25.7617, -80.1918),
    "Dallas": (32.7767, -96.7970)
}

initial_centroids = np.array([
    [40.0, -80.0], 
    [30.0, -100.0]
])

def euclidean_distance(point1, point2):
    return np.sqrt(np.sum((np.array(point1) - np.array(point2)) ** 2))

def map_step(cities, centroids):
    assignments = {}
    for city, coordinates in cities.items():
        closest_centroid = np.argmin([euclidean_distance(coordinates, c) for c in centroids])
        if closest_centroid not in assignments:
            assignments[closest_centroid] = []
        assignments[closest_centroid].append((city, coordinates))
    return assignments

def reduce_step(assignments):
    new_centroids = []
    for cluster, points in assignments.items():
        cluster_points = np.array([coords for _, coords in points])
        new_centroid = np.mean(cluster_points, axis=0)
        new_centroids.append(new_centroid)
    return np.array(new_centroids)

def plot_final_clusters(cities, centroids, assignments):
    colors = ['red', 'blue', 'green', 'purple']
    plt.figure(figsize=(8, 6))
    for cluster_id, points in assignments.items():
        cluster_points = np.array([coords for _, coords in points])
        plt.scatter(cluster_points[:, 1], cluster_points[:, 0], 
                    color=colors[cluster_id], label=f'Cluster {cluster_id}', alpha=0.6)
    plt.scatter(centroids[:, 1], centroids[:, 0], color='black', marker='x', s=100, label='Centroids')
    for city, (lat, lon) in cities.items():
        plt.text(lon, lat, city, fontsize=9, ha='right')
    plt.xlabel('Longitude')
    plt.ylabel('Latitude')
    plt.title('Final K-means Clustering Results')
    plt.legend()
    plt.grid(True)
    plt.show()

def kmeans(cities, initial_centroids, max_iterations=5):
    centroids = initial_centroids
    for iteration in range(max_iterations):
        print(f"\nIteration {iteration + 1}")
        assignments = map_step(cities, centroids)
        for cluster, points in assignments.items():
            print(f"Cluster {cluster}: {[city for city, _ in points]}")
        centroids = reduce_step(assignments)
        print(f"New Centroids: {centroids}")
    return centroids, assignments

final_centroids, final_assignments = kmeans(cities, initial_centroids)
plot_final_clusters(cities, final_centroids, final_assignments)
print("\nFinal Centroids:", final_centroids)
```
