```python
#Bigdata Lab prgm 2
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("mtcars.csv")

plt.hist(df['mpg'], bins=10, color='skyblue', edgecolor='black')

plt.title('Frequency Distribution of MPG (Miles per Gallon)')
plt.xlabel('MPG (Miles per Gallon)')
plt.ylabel('Frequency')
plt.show()
```

```python
# Bigdata Lab prgm 2
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("mtcars.csv")

# Set the plot style and background
plt.gca().set_facecolor('red')
plt.style.use('dark_background')

# Create the histogram
plt.hist(df['mpg'], color='skyblue', edgecolor='black')

# Set title and labels
plt.title('Frequency Distribution of MPG (Miles per Gallon)')
plt.xlabel('MPG (Miles per Gallon)')
plt.ylabel('Frequency')

# Customize the x-axis ticks
plt.xticks(range(10, int(df['mpg'].max()) + 10, 10))

# Show the plot
plt.show()
```
